```python
import asyncio
import os
import nest_asyncio
import uvicorn
import json
from fastapi import FastAPI, UploadFile, File, Form
from pyngrok import ngrok
import torch
from transformers import Qwen2_5_VLForConditionalGeneration, AutoProcessor
from qwen_vl_utils import process_vision_info

# 💡 Import Colab's secret manager
from google.colab import userdata

# 💡 FIX 1: Set environmental variable to reduce memory fragmentation
os.environ["PYTORCH_ALLOC_CONF"] = "expandable_segments:True"

print("⏳ Loading Qwen2.5-VL-3B-Instruct model onto Colab GPU...")
model_name = "Qwen/Qwen2.5-VL-3B-Instruct"

# Load model using bfloat16 to optimize memory usage
model = Qwen2_5_VLForConditionalGeneration.from_pretrained(
    model_name,
    torch_dtype=torch.bfloat16,
    device_map="auto"
)
processor = AutoProcessor.from_pretrained(model_name)
print("✅ Model successfully loaded on GPU!")

app = FastAPI()

@app.post("/analyze-video")
async def analyze_video(
    video: UploadFile = File(...),
    prompt: str = Form(...),
    generation_config: str = Form(None)  # 👈 Added to safely grab the custom settings sent from the bot script
):
    temp_path = "incoming_clip.mp4"
    with open(temp_path, "wb") as f:
        f.write(await video.read())

    try:
        # Default generation kwargs fallback
        # Настройки генерации по умолчанию
        gen_kwargs = {
            "max_new_tokens": 128,
            "use_cache": True,
            "do_sample": True,
            "temperature": 0.5,
            "top_p": 0.85,
            "repetition_penalty": 1.2
        }

        # Безопасно вытаскиваем настройки, пришедшие от бота
        if generation_config:
            try:
                config_dict = json.loads(generation_config)

                if "temperature" in config_dict:
                    gen_kwargs["temperature"] = float(config_dict["temperature"])
                    # Если temp = 0, то сэмплирование отключается (strict mode)
                    gen_kwargs["do_sample"] = True if gen_kwargs["temperature"] > 0 else False

                if "top_p" in config_dict:
                    gen_kwargs["top_p"] = float(config_dict["top_p"])

                if "repetition_penalty" in config_dict:
                    gen_kwargs["repetition_penalty"] = float(config_dict["repetition_penalty"])
                # Защита на случай, если бот всё ещё шлет старые penalty
                elif "presence_penalty" in config_dict or "frequency_penalty" in config_dict:
                    gen_kwargs["repetition_penalty"] = 1.25

                if "max_new_tokens" in config_dict:
                    gen_kwargs["max_new_tokens"] = int(config_dict["max_new_tokens"])

            except Exception as json_err:
                print(f"⚠️ Warning parsing generation_config JSON: {json_err}")

        # Enforce downsampling to protect against CUDA Out of Memory errors
        messages = [
            {
                "role": "user",
                "content": [
                    {
                        "type": "video",
                        "video": f"file://{os.path.abspath(temp_path)}",
                        "fps": 0.5,             # 1 frame per 2 seconds
                        "max_pixels": 360 * 360 # Rescale resolution to fit VRAM
                    },
                    {"type": "text", "text": prompt},
                ],
            }
        ]

        text = processor.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
        image_inputs, video_inputs = process_vision_info(messages)

        inputs = processor(
            text=[text],
            images=image_inputs,
            videos=video_inputs,
            padding=True,
            return_tensors="pt"
        ).to(model.device)

        with torch.inference_mode():
            # Pass the dynamic configurations inside the unpacked kwargs dict
            output_ids = model.generate(
                **inputs,
                **gen_kwargs
            )

        generated_ids_trimmed = [
            out_ids[len(in_ids):] for in_ids, out_ids in zip(inputs.input_ids, output_ids)
        ]
        response_text = processor.batch_decode(
            generated_ids_trimmed,
            skip_special_tokens=True,
            clean_up_tokenization_spaces=False
        )[0]

        return {"text": response_text.strip()}

    except Exception as e:
        return {"error": str(e)}

    finally:
        if os.path.exists(temp_path):
            os.remove(temp_path)
        # Force clear GPU memory cache after every video processing request
        torch.cuda.empty_cache()

# 💡 SAFE: Safely fetch the Ngrok token directly from your Colab Secrets drawer
try:
    NGROK_TOKEN = userdata.get('NGROK_AUTHTOKEN')
    ngrok.set_auth_token(NGROK_TOKEN)
except userdata.SecretNotFoundError:
    print("❌ Error: Could not find 'NGROK_AUTHTOKEN' in your Colab secrets drawer. Make sure the toggle is switched on!")
    raise

# Generate a tunnel link
public_url = ngrok.connect(8000)
print(f"\n🚀 COPY THIS PUBLIC URL FOR YOUR LOCAL SCRIPT:\n{public_url.public_url}\n")

# Start the API application loop safely inside Colab's event loop
nest_asyncio.apply()
config = uvicorn.Config(app, host="0.0.0.0", port=8000, loop="asyncio")
server = uvicorn.Server(config)
await server.serve()
```
